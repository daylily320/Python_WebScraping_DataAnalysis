```python
import requests
from bs4 import BeautifulSoup

url = 'https://news.daum.net/economy'

response = requests.get(url)

response.encoding = 'utf-8'

print(url)
print(type(response))
print(response.status_code)

html = response.text

soup = BeautifulSoup(html, 'html.parser')

li_tags = soup.select('ul.list_newsheadline2 li')

print(type(li_tags), len(li_tags))

for li_tag in li_tags:
    a_tag = li_tag.find('a')
    link = a_tag['href']

    strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
    title = strong_tag.text.strip()

    print(link)
    print(title)
```

```python
section_dict = {
    '기후/환경': 'climate',
    '사회': 'society',
    '경제': 'economy',
    '정치': 'politics',
    '국제': 'world',
    '문화': 'culture',
    '생활': 'life',
    'IT/과학': 'tech',
    '인물': 'people'
}

def print_news(section_name):
    section = section_dict[section_name]
    url = 'https://news.daum.net/' + section

    print(f'======> {url} {section_name} 뉴스 <======')

    response = requests.get(url)
    response.encoding = 'utf-8'

    soup = BeautifulSoup(response.text, 'html.parser')

    for li_tag in soup.select('ul.list_newsheadline2 li'):
        a_tag = li_tag.find('a')

        if a_tag is not None:
            link = a_tag['href']

            strong_tag = li_tag.select_one(
                'div.cont_thumb strong.tit_txt'
            )

            if strong_tag is not None:
                title = strong_tag.text.strip()

                print(link)
                print(title)
                print()


print_news('경제')
#print_news('사회')
```
